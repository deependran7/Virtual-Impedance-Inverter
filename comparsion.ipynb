```python
import pandas as pd
import matplotlib.pyplot as plt

plt.rcParams.update({
        "font.weight": "bold",  # bold fonts
        "font.size":12,
        "lines.linewidth": 2,   # thick lines
        "grid.color": "0.5",    # gray gridlines
        "grid.linewidth": 0.5,  # thin gridlines
        "savefig.dpi": 300,     # higher resolution output.
        "axes.labelsize" : 15,
        'legend.labelspacing':0.95,
    })
```

```python
tsimv,p1simv,p2simv,w1simv,w2simv,q1simv,q2simv,v1simv,v2simv,vav,vbv,vcv,iav,ibv,icv = pd.read_csv('Parallel Inverter Virtual Impedance\simu_data.csv').values.T
tsim,p1sim,p2sim,w1sim,w2sim,q1sim,q2sim,v1sim,v2sim,va,vb,vc,ia,ib,ic = pd.read_csv('Parallel Inverter Without Virtual Impedance\simu_data.csv').values.T
```

```python
xlim = [0.0,2]
fig, axs = plt.subplots(2, 2,figsize = (20,10))
axs[0, 0].plot(tsim,p1sim, label="Without Virtual Impedance")
axs[0, 0].plot(tsim,p1simv, label = "With Virtual Impedance")
axs[0, 0].plot(tsim,p2sim, label="Without Virtual Impedance")
axs[0, 0].plot(tsim,p2simv, label = "With Virtual Impedance")
axs[0, 0].grid()
axs[0, 0].set_xlim(xlim)
axs[0, 0].set_xlabel("Time")
axs[0, 0].set_ylabel("Active Power (W)")



axs[0,1].plot(tsim,q1sim,label="Without Virtual Impedance")
axs[0,1].plot(tsim,q1simv,label="With Virtual Impedance")
axs[0,1].plot(tsim,q2sim,label="Without Virtual Impedance")
axs[0,1].plot(tsim,q2simv,label="With Virtual Impedance")
axs[0,1].grid()
axs[0,1].set_xlim(xlim)
axs[0,1].set_xlabel("Time")
axs[0,1].set_ylabel("Reactive Power (Var)")


axs[1,0].plot(tsim,w1sim,label="Inv 1 Without Virtual Impedance")
axs[1,0].plot(tsim,w1simv,label="Inv 1 With Virtual Impedance")
axs[1,0].plot(tsim,w2sim,label="Inv 2 Without Virtual Impedance")
axs[1,0].plot(tsim,w2simv,label="Inv 2 With Virtual Impedance")
axs[1,0].grid()
axs[1,0].set_xlim(xlim)
axs[1,0].set_xlabel("Time")
axs[1,0].set_ylabel("Frequency (Hz)")
axs[1,0].legend(loc='upper center', bbox_to_anchor=(1, -0.2),fancybox=True, shadow=True, ncol=6)


axs[1,1].plot(tsim,v1sim,label="Without Virtual Impedance")
axs[1,1].plot(tsim,v1simv,label="With Virtual Impedance")
axs[1,1].plot(tsim,v2sim,label="Without Virtual Impedance")
axs[1,1].plot(tsim,v2simv,label="With Virtual Impedance")
axs[1,1].grid()
axs[1,1].set_xlim(xlim)
axs[1,1].set_xlabel("Time")
axs[1,1].set_ylabel("Voltage, $v_{od}$ (V)")
plt.savefig('comparsion.jpg',dpi = 300, bbox_inches='tight')
```
